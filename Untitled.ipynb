```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('titanic.csv')
print(f'Размер датасета: {df.shape[0]} строк, {df.shape[1]} столбцов')
display(df.head(10))
df.info()

print('\n=== Пропущенные значения ДО заполнения ===')
print(df.isnull().sum())

age_median = df['Age'].median()
cabin_mode = df['Cabin'].mode()[0]
embarked_mode = df['Embarked'].mode()[0]
print(f'\nМедиана Age: {age_median}')
print(f'Мода Cabin: {cabin_mode}')
print(f'Мода Embarked: {embarked_mode}')

df['Age'].fillna(age_median, inplace=True)
df['Cabin'].fillna(cabin_mode, inplace=True)
df['Embarked'].fillna(embarked_mode, inplace=True)

print('\n=== Пропущенные значения ПОСЛЕ заполнения ===')
print(df.isnull().sum())

df = df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis='columns')

numeric_cols = df.select_dtypes(include='number').columns.drop('Survived')
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
print('\n=== Данные после нормализации ===')
display(df.head(10))

object_cols = df.select_dtypes(include='object').columns.tolist()
print(f'\nКатегориальные столбцы: {object_cols}')
df = pd.get_dummies(df, columns=object_cols, drop_first=True)
print('\n=== Финальный датасет после OHE ===')
display(df.head(10))

train_df, test_df = train_test_split(df, test_size=0.3, random_state=42)
print(f'\nОбучающая выборка: {train_df.shape}')
print(f'Тестовая выборка: {test_df.shape}')

df.to_csv('processed_titanic.csv', index=False)
print('\nОбработанный датасет сохранён в processed_titanic.csv')
```
